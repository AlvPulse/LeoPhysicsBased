```python
import os

# ⚠️ REPLACE THIS WITH YOUR FOLDER PATH
folder_path = "C:/Users/Mostafa/Downloads/TAU-urban-acoustic-scenes-2020-3class-evaluation.audio.13/TAU-urban-acoustic-scenes-2020-3class-evaluation/audio/Kitchen"

# Get all files (ignore subdirectories)
files = [f for f in os.listdir(folder_path) 
         if os.path.isfile(os.path.join(folder_path, f))]

# Sort files for consistent ordering (optional but recommended)
files.sort()

counter = 1
for file in files:
    # Generate new filename (e.g., "abc1.wav")
    new_name = f"Kitchen{counter}.wav"
    
    # Full paths
    old_path = os.path.join(folder_path, file)
    new_path = os.path.join(folder_path, new_name)
    
    # Rename the file
    os.rename(old_path, new_path)
    counter += 1

print(f"Renamed {len(files)} files successfully!")
```
